```python
import sys
from pathlib import Path
from researchcodes import (
    define_column_desc, 
    iter_multi_csv_chunks, 
    write_std_h5, 
)
```

# Define the HFD5 column description

```python
# Define magnitude column names

# Note this magnitude column order is not necessarily
# to be the same as the column order in the text file. 

# It is OK as long as the filter names in 
# `magnitude_column_names` matches `colnames` defined 
# when reading the text file

magnitude_column_names = [
    "Johnson_B", "Johnson_V",
    "SDSS_up", "SDSS_gp", "SDSS_rp", "SDSS_ip",
    "PanSTARRS_zs","PanSTARRS_Y",
    "Johnson_B_err", "Johnson_V_err",
    "SDSS_up_err", "SDSS_gp_err", "SDSS_rp_err", "SDSS_ip_err",
    "PanSTARRS_zs_err","PanSTARRS_Y_err",
]

# define h5 file column description
h5_columns = define_column_desc(
    magnitude_column_names=magnitude_column_names, 
    id_name_length=16, 
)
```

```python
h5_columns
```

# Read the cvs files

```python
# list all the files
root_dir = Path(".")
files = list(root_dir.glob("*.txt"))
#files
```

```python
# csv column names
colnames = [
    "id_name", "ra", "ra_err", "dec", "dec_err",
    "nobs_B","nobs_V","nobs_u","nobs_g","nobs_r","nobs_i","nobs_PanSTARRS_zs","nobs_PanSTARRS_Y",
    "Johnson_B", "Johnson_V",
    "SDSS_up", "SDSS_gp", "SDSS_rp", "SDSS_ip",
    "PanSTARRS_zs","PanSTARRS_Y",
    "Johnson_B_err", "Johnson_V_err",
    "SDSS_up_err", "SDSS_gp_err", "SDSS_rp_err", "SDSS_ip_err",
    "PanSTARRS_zs_err","PanSTARRS_Y_err",
]

dataframe_iterator = iter_multi_csv_chunks(
    files=files, 
    chunksize=100, 
    read_csv_kwargs={
        "sep": r"\s+", 
        "engine": "python", 
        "header": None, 
        "names": colnames,
        "skiprows": 1, 
    }
)
```

```python
table_attrs = {
    "source": "Contact aavso@aavso.org.",
    "ra_unit": "deg", 
    "dec_unit": "deg", 
    "ra_err_unit": "arcsec", 
    "dec_err_unit": "arcsec", 
    "version": "dr10",
    "mag_system": {
        "Johnson_BV": "Vega", 
        "SDSS_ugri_primed": "AB", 
        "PanSTARRS_zsY": "AB", 
    }, 
}
```

```python
write_std_h5(
    dataframe_iterator=dataframe_iterator, 
    ra_dec_hmsdms=False,
    h5_output_path=Path("APASSDR10.h5", ), 
    group_where="/", 
    group_name="apass", 
    group_title="APASS catalogs", 
    table_name="std", 
    table_description=h5_columns, 
    table_title="Standard Stars", 
    table_attrs=table_attrs, 
    nside=512, 
    bucket_size=1536 , 
)
```
