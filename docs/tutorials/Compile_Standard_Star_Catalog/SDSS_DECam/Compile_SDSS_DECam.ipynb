```python
import pandas as pd
import numpy as np
from pathlib import Path
from researchcodes import (
    define_column_desc, 
    iter_multi_csv_chunks, 
    write_std_h5, 
)
```

```python
# Download full SDSS DECam standard star catalog from
# https://noirlab.edu/science/programs/ctio/instruments/Dark-Energy-Camera/User-Guide/Photometric-Standard-Stars#SDSS
```

# Define the HFD5 column description

```python
# Define magnitude column names

# Note this magnitude column order is not necessarily
# to be the same as the column order in the text file. 

# It is OK as long as the filter names in 
# `magnitude_column_names` matches `colnames` defined 
# when reading the text file

magnitude_column_names = [
    "SDSS_up", "SDSS_gp", "SDSS_rp", "SDSS_ip", "SDSS_zp", "UKIDSS_LAS_Y",
    "SDSS_up_err", "SDSS_gp_err", "SDSS_rp_err", "SDSS_ip_err", "SDSS_zp_err", "UKIDSS_LAS_Y_err",
]

# define h5 file column description
h5_columns = define_column_desc(
    magnitude_column_names=magnitude_column_names, 
    id_name_length=30, 
)
```

```python
h5_columns
```

# Read the cvs files

```python
# file path
root_dir = Path(".")
file = list(root_dir.glob("SDSS*csv"))
file
```

```python
# csv column names
colnames = [
    "id_name",
    "ra",
    "dec",
    "SDSS_up", "SDSS_up_err",
    "SDSS_gp", "SDSS_gp_err",
    "SDSS_rp", "SDSS_rp_err",
    "SDSS_ip", "SDSS_ip_err",
    "SDSS_zp", "SDSS_zp_err",
    "UKIDSS_LAS_Y", "UKIDSS_LAS_Y_err",
]

dataframe_iterator = iter_multi_csv_chunks(
    files=file, 
    chunksize=1000, 
    read_csv_kwargs={
        "sep": ",", 
        "engine": "python", 
        "header": None, 
        "names": colnames,
        "skiprows": 1, 
    }
)
```

```python
table_attrs = {
    "source": "https://www-star.fnal.gov/DECamStds/",
    "ra_unit": "deg",
    "dec_unit": "deg",
    "ra_err_unit": "NaN",
    "dec_err_unit": "NaN",
    "version": "v1.0",
    "mag_system": {
        "SDSS_ugriz": "AB",
        "UKIDSS_LAS_Y": "AB"
    },
}
```

```python
write_std_h5(
    dataframe_iterator=dataframe_iterator, 
    h5_output_path=Path("SDSS_DECam.h5",), 
    group_where="/sdss", 
    group_name="decam", 
    group_title="DeCam", 
    table_name="std", 
    table_description=h5_columns, 
    table_title="Standard Stars", 
    table_attrs=table_attrs, 
    nside=512, 
    bucket_size=1536 , 
)
```
