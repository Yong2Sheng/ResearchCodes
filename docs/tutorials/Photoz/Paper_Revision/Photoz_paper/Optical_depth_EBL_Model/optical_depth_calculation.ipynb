# Calculate the how accuracy photo-z can constrain the horizon

```python
# This gets horizon energy when you know z and tau

import numpy as np
from scipy.interpolate import InterpolatedUnivariateSpline

def tau_of(z, E):
    """Return τ(z, E) as a float."""
    return float(interp(z, E)[0][0])   # interp returns a 1x1 array

def horizon_energy(z0, tau_target):
    """
    Find the energy E such that tau(z0, E) = tau_target.
    Uses a 1D spline in energy at fixed redshift.
    """
    # τ as a function of energy for this redshift
    tau_vals = np.array([tau_of(z0, E) for E in energies])

    # Build a spline for f(E) = τ(z0,E) - tau_target
    f_spline = InterpolatedUnivariateSpline(energies, tau_vals - tau_target)

    # Find the roots (where f(E) = 0)
    roots = f_spline.roots()
    if len(roots) == 0:
        raise ValueError(f"No E found with tau={tau_target} at z={z0}")
    # If there's more than one, pick the first (or choose by energy range)
    return roots[0]
```

```python
def find_z_true(z, eta):
    lower = (z - eta) / (1 + eta)
    upper = (z + eta) / (1 - eta)
    # Optional: ensure lower <= upper
    if lower > upper:
        lower, upper = upper, lower
    return lower, upper

def find_delta_tau(z, tau_target, eta):

    energy = horizon_energy(z, tau_target)

    lower_z_true, upper_z_true = find_z_true(z, eta)

    lower_z_true_tau = float(interp(lower_z_true, energy)[0][0])
    upper_z_true_tau = float(interp(upper_z_true, energy)[0][0])

    max_delta_tau = np.max([
        np.abs(tau_target - lower_z_true_tau),
        np.abs(tau_target - upper_z_true_tau)
    ])

    fractional_error = (upper_z_true_tau-lower_z_true_tau)/(2*tau_target)

    print(f"Photometric redshift z_phot = {z} implies true redshift in [{lower_z_true:.3f}, {upper_z_true:.3f}] for |η| = {eta}.")
    print(f"At E = {energy:.4g} TeV, tau(z_phot) = {tau_target}.")
    print(f"Lower true redshift has tau = {lower_z_true_tau:.4f}.")
    print(f"Upper true redshift has tau = {upper_z_true_tau:.4f}.")
    print(f"Max |Δtau| = {max_delta_tau:.4f}")
    print(f"The symmetric fractional error is {fractional_error}")
    

    return lower_z_true, upper_z_true, max_delta_tau, fractional_error
```
