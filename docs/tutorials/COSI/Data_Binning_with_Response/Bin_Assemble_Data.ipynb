# Some imports

```python
from pathlib import Path
from histpy import Histogram
import matplotlib.pyplot as plt
from astropy.coordinates import SkyCoord
import astropy.units as u
import numpy as np
from cosipy import TransientBackgroundEstimation
from cosipy import SpacecraftHistory
from astropy.time import Time
```

```python
def set_binning_from_response(analysis, response_path):
    """Update a BinnedData object in place using response binning."""

    with FullDetectorResponse.open(response_path) as response:
        analysis.energy_bins = (
            response.axes["Em"]
            .edges
            .to_value(u.keV)
            .tolist()
        )

        phi_edges = response.axes["Phi"].edges.to_value(u.deg)
        phi_widths = np.diff(phi_edges)

        if not np.allclose(phi_widths, phi_widths[0]):
            raise ValueError(
                "The response has nonuniform Phi bins, but BinnedData "
                "requires a single phi_pix_size."
            )

        if not (
            np.isclose(phi_edges[0], 0.0)
            and np.isclose(phi_edges[-1], 180.0)
        ):
            raise ValueError(
                "The response Phi axis does not cover 0–180 degrees."
            )

        analysis.phi_pix_size = float(phi_widths[0])

        psichi_axis = response.axes["PsiChi"]
        analysis.nside = int(psichi_axis.nside)
        analysis.scheme = str(psichi_axis.scheme).lower()
```

# Major inputs

```python
GRB_name = "GRB_bn080802386"

raw_signal_file_name = "GRB_bn080802386_flux150_3months_unbinned_data_filtered_with_SAAcut.fits.gz"

# This the true location of the GRB
coord = SkyCoord(l = 23.53, b = -53.44, unit = (u.deg, u.deg), frame = "galactic")

# this is the unbinned GRB start and end times
GRB_start_time_unbinned = 1835493492.2267933
GRB_end_time_unbinned =   1835493492.7676215

# this is the start and end of GRBs
GRB_start_time = np.floor(GRB_start_time_unbinned)
GRB_end_time = np.ceil(GRB_end_time_unbinned)

# bining options
time_bin_size = 0.1
```

# Set up file paths

```python
from pathlib import Path

data_dir = Path("/project/majello/astrohe/yong/COSI/Data_Challenges/DC4")

GRB_unbinned_path = data_dir / raw_signal_file_name

orientation_path = data_dir / "DC4_final_530km_3_month_with_slew_1sbins_GalacticEarth_SAA.fits"

response_path = data_dir / "ResponseContinuum.o4.e100_10000.b10log.s10396905069491.m166.filtered.nonsparse.binnedimaging.imagingresponse_nside16.area.h5"

bkg_unbinned_path = data_dir / "Total_DC4_BG_3months_unbinned_data_filtered_with_SAAcut_withSAAbck.fits.gz"
```

```python
ori = SpacecraftHistory.open(orientation_path)

if 0 in ori.select_interval(
    Time(GRB_start_time, format="unix"),
    Time(GRB_end_time, format="unix"),
).livetime:
    print(f"Satellite shut down during GRB {GRB_name}")
```

# Decide binning time range

```python
# Define the time range used to bin and assemble the GRB signal and background.
# A long background interval is required to obtain a more stable background
# estimate, since the current background model is not sufficiently smooth.
#data_start_time = GRB_start_time - 500
#data_end_time = GRB_end_time + 500

data_start_time = GRB_start_time_unbinned - 500
data_end_time = GRB_end_time_unbinned + 500
```

# Bin signal

```python
from cosipy import BinnedData

import logging
logging.basicConfig(level=logging.INFO)

%matplotlib inline 
```

```python
analysis = BinnedData("inputs.yaml")

analysis.tmin = data_start_time
analysis.tmax = data_end_time

analysis.time_bins = time_bin_size

set_binning_from_response(
    analysis,
    response_path,
)
```

```python
%%time

analysis.get_binned_data(
    unbinned_data=GRB_unbinned_path,
    output_name=data_dir / f"{GRB_name}_signal_binned",
    psichi_binning="local",
)
```

```python
# check signal light curve
%matplotlib inline

signal = Histogram.open(data_dir / f"{GRB_name}_signal_binned.hdf5")
estimator = TransientBackgroundEstimation(signal)

# view the full light curve
estimator.plot_lightcurve()
```

```python
# get GRB signal counts
signal_times = signal.axes['Time'].edges.value

GRB_tmin_idx = np.searchsorted(signal_times, GRB_start_time, side='left')
GRB_tmax_idx = np.searchsorted(signal_times, GRB_end_time, side='right')

grb_singal = signal.slice[GRB_tmin_idx:GRB_tmax_idx,:]  # It slices the Time axis

grb_counts = np.sum(grb_singal)
grb_counts
```

# Bin true background

```python
from cosipy import BinnedData

import logging
logging.basicConfig(level=logging.INFO)

%matplotlib inline 
```

```python
analysis = BinnedData("inputs.yaml")

analysis.tmin = data_start_time
analysis.tmax = data_end_time

analysis.time_bins = time_bin_size

set_binning_from_response(
    analysis,
    response_path,
)
```

```python
%%time

analysis.get_binned_data(
    unbinned_data=bkg_unbinned_path,
    output_name=data_dir / f"{GRB_name}_bkg_binned",
    psichi_binning="local",
)
```

```python
# check bkg light curve
%matplotlib inline

bkg = Histogram.open(data_dir / f"{GRB_name}_bkg_binned.hdf5")
estimator = TransientBackgroundEstimation(bkg)

# view the full light curve
estimator.plot_lightcurve()
```

```python
# background counts
bkg_times = bkg.axes['Time'].edges.value

GRB_tmin_idx = np.searchsorted(bkg_times, GRB_start_time, side='left')
GRB_tmax_idx = np.searchsorted(bkg_times, GRB_end_time, side='right')

grb_bkg = bkg.slice[GRB_tmin_idx:GRB_tmax_idx,:]  # It slices the Time axis

bkg_counts = np.sum(grb_bkg)
bkg_counts
```

```python
# compuate S/N
snr_total = grb_counts / np.sqrt(grb_counts + bkg_counts)
snr_bkg = grb_counts / np.sqrt(bkg_counts)
signal_fraction = grb_counts / (grb_counts + bkg_counts)

print("S/sqrt(S+B):", snr_total)
print("S/sqrt(B):", snr_bkg)
print("Signal fraction:", signal_fraction)
```
