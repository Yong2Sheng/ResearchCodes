```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```python
pl_QSO_DR2_029_t0_fname = "pl_QSO_DR2_029_t0.spec"
S0_70_QSO2_30_fname  = "S0_70_QSO2_30.sed"
pl_QSOH_template_norm_fname = "pl_QSOH_template_norm.sed"
```

```python
df_pl_QSO_DR2_029_t0 = pd.read_csv(pl_QSO_DR2_029_t0_fname, sep='\s+', names=["WL", "Fv"])
df_S0_70_QSO2_30 = pd.read_csv(S0_70_QSO2_30_fname, sep='\s+', names=["WL", "Fv"])
df_pl_QSOH_template_norm = pd.read_csv(pl_QSOH_template_norm_fname, sep='\s+', names=["WL", "Fv"])
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(16,8), sharex=False, nrows=1) ##sharex=True

ax.plot(df_pl_QSO_DR2_029_t0["WL"], df_pl_QSO_DR2_029_t0["Fv"], label = pl_QSO_DR2_029_t0_fname)
ax.plot(df_S0_70_QSO2_30["WL"], df_S0_70_QSO2_30["Fv"], label = S0_70_QSO2_30_fname)
ax.plot(df_pl_QSOH_template_norm["WL"], df_pl_QSOH_template_norm["Fv"], label = pl_QSOH_template_norm_fname)
ax.axvspan(4000, 8000, alpha=0.5, color='gray')

plt.yscale("log")
plt.xscale("log")

plt.legend(fontsize = 20)

plt.tick_params(axis="both", which="both", labelleft=True, labelright=True, labelbottom=True, labeltop=True, labelsize = 20)

plt.xticks(fontsize=20)
plt.yticks(fontsize=20);
```
