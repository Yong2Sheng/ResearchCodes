# Calculate MgII none Detection Probability

```python
import numpy as np

def mg2_prob(dndz, dndz_err, deltaz, decimals=1):
    """
    dndz: central value
    dndz_err: (upper, lower) e.g. (0.22, 0.14) means +0.22/-0.14
    deltaz: redshift path
    returns: (P_cen, P_plus, P_minus, latex_string)
    """
    up, low = dndz_err  # upper, lower (both positive numbers)

    # expected counts
    mu_c = dndz * deltaz
    mu_hi = (dndz + up) * deltaz   # more absorbers -> LOWER probability
    mu_lo = (dndz - low) * deltaz  # fewer absorbers -> HIGHER probability

    # probabilities (in %)
    P_c = 100.0 * np.exp(-mu_c)
    P_low = 100.0 * np.exp(-mu_hi)  # lower bound of P
    P_high = 100.0 * np.exp(-mu_lo) # upper bound of P

    # asymmetric errors
    dP_plus = P_high - P_c
    dP_minus = P_c - P_low

    # rounding
    P_c_r = np.round(P_c, decimals)
    dP_plus_r = np.round(dP_plus, decimals)
    dP_minus_r = np.round(dP_minus, decimals)

    latex = f"{P_c_r}^{{+{dP_plus_r}}}_{{-{dP_minus_r}}}\\%"

    return P_c_r, dP_plus_r, dP_minus_r, latex
```

```python
# 示例：dndz=0.33, ( +0.22, -0.14 ), Δz=1.07
print(mg2_prob(0.23, (0.13, 0.09), 1.502)[-1])
```
