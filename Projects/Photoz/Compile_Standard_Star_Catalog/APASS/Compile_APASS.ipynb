```python
import sys
from pathlib import Path

# set up the system path to import the 
# compile_standard_catalog module 
extra = Path("..").resolve()
sys.path.insert(0, str(extra))

from compile_standard_catalog import (
    define_column_desc, 
    iter_multi_csv_chunks, 
    write_std_h5, 
)


```

# Define the HFD5 column description

```python
magnitude_column_names = [
     "mag_B","mag_V","mag_u","mag_g","mag_r","mag_i","mag_PanSTARRS_zs","mag_PanSTARRS_Y",
    # emag 8个
    "emag_B","emag_V","emag_u","emag_g","emag_r","emag_i","emag_PanSTARRS_zs","emag_PanSTARRS_Y",
]

apass_h5_columns = define_column_desc(
    magnitude_column_names=magnitude_column_names, 
    id_name_length=16, 
)
```

# Read the cvs files

```python
# list all the files
root_dir = Path(".")
files = list(root_dir.glob("*.txt"))
#files
```

```python
# csv column names
colnames = [
    "id_name", "ra", "ra_err", "dec", "dec_err",
    # nobs 8个
    "nobs_B","nobs_V","nobs_u","nobs_g","nobs_r","nobs_i","nobs_PanSTARRS_zs","nobs_PanSTARRS_Y",
    # mag 8个
    "mag_B","mag_V","mag_u","mag_g","mag_r","mag_i","mag_PanSTARRS_zs","mag_PanSTARRS_Y",
    # emag 8个
    "emag_B","emag_V","emag_u","emag_g","emag_r","emag_i","emag_PanSTARRS_zs","emag_PanSTARRS_Y",
]

dataframe_iterator = iter_multi_csv_chunks(
    files=files, 
    chunksize=100, 
    read_csv_kwargs={
        "sep": r"\s+", 
        "engine": "python", 
        "header": None, 
        "names": colnames,
        "skiprows": 1, 
    }
)
```

```python
table_attrs = {
    "ra_unit": "deg", 
    "dec_unit": "deg", 
    "ra_err_unit": "arcsec", 
    "dec_err_unit": "arcsec", 
    "mag_system": {
        "Johnson_BV": "Vega", 
        "SDSS_ugri": "AB", 
        "PanSTARRS_zsY": "AB", 
    }, 
}
```

```python
write_std_h5(
    dataframe_iterator=dataframe_iterator, 
    h5_output_path=Path("test.h5", ), 
    group_where="/", 
    group_name="apass", 
    group_title="APASS catalogs", 
    table_name="dr10", 
    table_description=apass_h5_columns, 
    table_title="APASS DR10", 
    table_attrs=table_attrs, 
    nside=512, 
    bucket_size=1536 , 
)
```
