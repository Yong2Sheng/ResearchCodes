```python
import sys
from pathlib import Path
from tqdm.notebook import tqdm

# set up the system path to import the plot_spec_file module
# It means the relative structure between LePHARE_Fitting_Fortran
# and Plot_Spec_File folder can't change.
extra = Path("../Plot_Spec_File").resolve()
sys.path.insert(0, str(extra))

from plot_spec_file import SpecResults
```

# Plot galaxy results

```python
root_dir = Path("galaxy_model_fitting").resolve()
file_list = list(root_dir.glob("I*.spec"))
```

```python
for file in tqdm(file_list):

    save_path = root_dir / f"{file.stem}.png"

    print(f"Workong on {save_path}")

    spec = SpecResults(file)
    spec.plot_results(save_path = save_path)
```
